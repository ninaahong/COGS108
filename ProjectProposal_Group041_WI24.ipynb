# Research Question

How does music preference correlate with college major choice among UCSD students, and can this correlation be reflected in the popularity of music artists among different majors? What is the relationship between students' music taste, measured by their preferred artists or genres, and their chosen field of study, with a focus on identifying any patterns or trends in music preference across various majors? 

## Background and Prior Work

Choosing a major is one of the most important decisions college students make. This choice impacts their future careers and their development as a person. There are many factors that one considers when taking into account what they want to study such as high school classes interests, what their family members studied, and many more. For our project we want to investigate if there are any subconscious influences when choosing a major. To do so, we want to see if there is a correlation between a student's music taste and major election. 

A study pertaining to music taste and personality showed that people with certain personality traits tend to enjoy certain music genres more than others <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1). This study showed that music is somehow embedded in certain personality traits of people. Similarly, other studies found out that personality traits do play a role when choosing a college major <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). Both of these studies are making reference to the “Big Five Personality Traits”. By seeing that personality is related to both music and major choice, we want to discover the relationship between both or if it is solely the personality and music taste that is not really reflected among majors.

The relationship between music taste and other life aspects has been studied before as well. For instance, a study looked at the relationship between music taste and job choice and they found there was a relation between both <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). This research study is important because it shows that music can be related to people’s life and career so based on this we want to explore how music plays a role before choosing a job.

# Hypothesis


We predict that there will be a correlation between music preference and college major choice among UCSD students. Building on previous research indicating connections between personality traits and both music preference and major selection, we hypothesize that certain music genres may resonate more with the personality traits associated with specific majors. Therefore, we expect to observe a pattern where certain music artists or genres are more popular among students majoring in particular fields, reflecting underlying personality traits influencing both music preference and major choice.

# Data

Since there is no existent dataset that could help us in addressing our research question, we decided to collect the data by ourselves. 

In order to effectively address our research question regarding the relationship between music preference and college majors among UCSD students, an ideal dataset should include the following variables: 
1. Major of choice: The particular major or program that the student is enrolled in.
2. Music preference: Information on the types of music that the student finds enjoyable.
3. Artist preference: Information on the specific artists/bands the student enjoys listening to.

We aim for a sample size that is representative of the UCSD student population to ensure our findings can be generalized. This indicates that there should be over 100 observations in our sample. 

Considering that UCSD students are our target population, we would use an online survey to collect our data. We could create a Google Form or a similar platform to host the questionnaire. Subsequently, the survey's link would be shared via various other channels, including student forums like Piazza and social media sites like Instagram. This method would facilitate participation and accessibility, enabling us to gather a diverse range of responses.

Responses collected from the survey would be stored in a structured format, such as a spreadsheet. Every row in the spreadsheet denotes a participant, and every column denotes a variable. This organized dataset would facilitate data analysis and interpretation.

# Ethics & Privacy

The collection of students' music preferences paired with their majors involves handling personal information, which raises privacy concerns. Another potential issue is obtaining consent from students, where we need to guarantee that students will remain untraceable and anonymous, ensuring that responses cannot be traced back to individuals without their explicit permission. There is a risk of self-selection bias, where individuals who have a particular interest in music or strong opinions about their preferences may be more likely to respond. This could lead to an overrepresentation of certain preferences, which may not be reflective of the general student population at UCSD. Our approach to detecting specific biases before communicating our analysis includes a comprehensive evaluation of our data sources. Conducting this evaluation ensures that we're working with a dataset that's both diverse and free from inherent biases, setting a strong foundation for unbiased analysis. Throughout the analytical process, we’ll engage in ongoing validation efforts against varied data subsets and perform detailed sensitivity analyses. These efforts aim to understand how assumptions may influence our findings, helping us ensure our results are reflective of reality. Integrating this feedback into our work is essential for boosting the reliability of our analysis outcomes and demonstrating our commitment to maintaining high ethical standards.

To tackle identified issues, we will first conduct a root cause analysis, as previously mentioned, to understand the underlying problems and determine whether they stem from data quality, bias, or methodological flaws. Then, to address these challenges, we’ll use tactics that involve methodological adjustments and data remediation to ensure that our analysis is accurate. Additionally, using various methods beyond digital platforms, such as flyers, will help us reach a more diverse audience and address our concerns about bias. By being transparent with participants about how their data will be used and providing clear information on the purpose of the survey, we will help ensure that students' consent is informed and voluntary. Implementing these tactics will allow us to pinpoint more efficiently where issues stem from and facilitate our continuous improvement.

The standardize_age function below takes a string input which represents an age value and standardizes it by removing any additional text (e.g. 'years old' or 'years') and returning the age as an integer. This code ensures that the 'age' column in the 'survey' DataFrame is cleaned and standardized, making it ready for further analysis or visualization as it will only be an integer. 

To figure out how to standardize the age responses, we manually looked through the entries to sort out what we need to replace. We only wanted the entries to be integers that correlated with age, so we removed all texts.

```python
##MAKE THE AGE COLUMN AN INTEGER
def standardize_age(str_age):
    
    str_age = str(str_age)
    str_age = str_age.lower()
    str_age = str_age.strip()

    if 'years old' in str_age:
        str_age = str_age.replace('years old', '')
        output = str_age
    elif 'years' in str_age:
        str_age = str_age.replace('years', '')
        output = str_age
    else:
        output = str_age
    return int(output)
    
survey['age'] = survey['age'].apply(standardize_age)
survey['age'] = survey['age'].astype(int)
```

We want to ensure consistency and clarity in the representation of students' favorite music genres in this survey dataset. We want to convert all genre labels to lowercase and standardizing certain genre names.

Sometimes, the names may contain hypgens or other marks and we should remove them to treat genres as single entities. This standardization process simplifies the data, ensuring that similar genres are grouped together consistently for analysis.

To clean the genre columns up a bit, we needed to turn all responses into lower case and remove any leading whitespace. Then, we manually looked through the entries and decided that we need to remove symbols and generalize some genres.

```python
##FAVORITE GENRE DATA CLEANING
def standardize_gen(str_gen):

    str_gen = str(str_gen)
    str_gen = str_gen.lower()
    str_gen = str_gen.strip()

    if '-' in str_gen:
        str_gen = str_gen.replace('-', '')
        output = str_gen
    elif 'heavy metal' in str_gen or 'power metal' in str_gen:
        output = 'metal'
    elif 'alternative rock' in str_gen:
        output = 'rock'
    elif 'oldies' in str_gen:
        output = np.nan
    else:
        output = str_gen
    return output
    
survey['favorite_genre_1'] = survey['favorite_genre_1'].apply(standardize_gen)
survey['favorite_genre_2'] = survey['favorite_genre_2'].apply(standardize_gen)
survey['favorite_genre_3'] = survey['favorite_genre_3'].apply(standardize_gen)
```

There seems to be more symbols and phrases such as 'less than', '+', '~', and '?', as well as variations of 'hour(s)', so let's ensure uniform formatting of the hour values. Also, we need to assign midpoints for specified ranges that cannot be decided as a single value.

To clean up the entries, we first removed any symbols and texts since we wanted the entries in the listening hours column to be floats. We manually looked through the entries to decide what we needed to replace.

```python
##CLEAN HOURS BY REMOVING SYMBOLS AND ADDING AVERAGES FOR PEOPLE WHO PROVIDED RANGE
def standardize_hours(str_hours):
    
    str_hours = str(str_hours)
    str_hours = str_hours.lower()
    str_hours = str_hours.strip()

    if 'less' in str_hours or '+' in str_hours or ' hour' in str_hours or ' hrs ' in str_hours or '~' in str_hours or '?' in str_hours:
        str_hours = str_hours.replace('less than ', '')
        str_hours = str_hours.replace('+', '')
        str_hours = str_hours.replace(' hours', '')
        str_hours = str_hours.replace(' hour', '')
        str_hours = str_hours.replace(' hrs ', '')
        str_hours = str_hours.replace('~', '')
        str_hours = str_hours.replace('?', '')
        output = str_hours       
    else:
        output = str_hours
    return output

survey['listening_hours'] = survey['listening_hours'].apply(standardize_hours)

def standardize_avg(str_avg):
    
    str_avg = str(str_avg)
    
    if'0-1'in str_avg:
        output = '0.5'
    elif '1-2' in str_avg:
        output = '1.5'
    elif '2-3' in str_avg:
        output = '2.5'
    elif '3-4' in str_avg:
        output = '3.5'
    elif '5-7' in str_avg:
        output = '6'
    elif ' ' in str_avg:
        str_avg = str_avg.replace('4 ', '4')
        str_avg = str_avg.replace('3 ', '3')
        output = str_avg
    elif '1/3' in str_avg:
        output = np.nan
    else:
        output = str_avg
    return output
    
survey['listening_hours'] = survey['listening_hours'].apply(standardize_avg)
survey['listening_hours'] = survey['listening_hours'].astype(float)
```

```python
artists_genre['names'] = artists_genre['names'].str.lower()
artists_genre[['name','genre']] = artists_genre['names'].str.split(':',expand = True)
artists_genre = artists_genre.drop(['names'],axis = 1)
artists_genre['genre'].unique()

def standardize_genre (str_g):
    
    str_g = str(str_g)
    str_g = str_g.lower()
    str_g = str_g.strip()
    
    if '-' in str_g:
        str_g = str_g.replace('-','')
        output = str_g
    elif 'electronic' in str_g:
        output = 'edm'
    elif 'contemporary christian' in str_g:
        output = 'christian'
    elif 'psychedelic pop' in str_g:
        output = 'pop'
    elif 'power metal' in str_g or 'progressive metal' in str_g or 'alternative metal' in str_g:
        output = 'metal'
    elif 'hard rock' in str_g or 'pop rock' in str_g or 'alternative rock' in str_g or 'folk rock' in str_g:
        output = 'rock'
    elif 'pop punk' in str_g:
        output = 'punk'
    elif 'indie rock' in str_g or 'indie pop' in str_g:
        output = 'indie'
    elif 'cloud rap' in str_g:
        output = 'hiphop/rap'
    else:
        output = str_g
    return output

artists_genre['genre'] = artists_genre['genre'].apply(standardize_genre)
artists_genre['genre'] = artists_genre['genre'].replace('hiphop', 'hiphop/rap')

artists_genre.head()
```

### Chi test for Major-Genre Relation

Now, we are going to conduct a chi2 test to see if these two variables, 'major' and 'genre' are independent or not to understand if there is a relationship between them or not. We will decide this based on the value of the p_value.

```python
chi, p_value, degree, expected = chi2_contingency(contigency_table) 
if p_value > 0.05:
     print('Fail to reject null')
else:
     p_value <= 0.05
     print('Reject null')
```

Given that we should fail to reject the null hypothesis, it seems there is no correlation between major and music taste for UCSD students

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
## FEEL FREE TO ADD MULTIPLE CELLS PER SECTION 

##MAKE SURE TO INCLUDE HERE ANY OTHER IMPORTS YOU MIGHT NEED WE CAN NEED
import pandas as pd
import numpy as np

survey=pd.read_csv('cogs - 108 survey (Responses) - Form Responses 1.csv')
```

```python
##DROP COLUMNS WE WILL NOT  USE
survey=survey.drop(['Timestamp'],axis=1)
survey=survey.drop(['What is/are your favorite color(s)? '],axis=1)
survey=survey.drop(["What are your hobbies? If it's more than one please put commas in between your responses. "],axis=1)
```

```python
##RENAME COLUMNS
survey = survey.rename(columns={'Age':'age',
                                'Gender':'gender',
                                'Hometown (include only state and city)': 'state_country', 
                                'favorite genre #1':'favorite_genre_1',
                                'favorite genre #2':'favorite_genre_2',
                                'favorite genre #3':'favorite_genre_3',
                                'Major/majors':'major_majors',
                                'Who is your favorite artist(s)?':'favorite_artist',
                                'How many hours per day do you listen to music? ': 'listening_hours',
                                'Do you play any instrument': 'instrument',
                                'What instrument do you play?': 'instrument_played',
                               'Are you an introvert or extrovert? ':'introvert_extrovert'})
```

```python
##REMOVE LETTERS FROM AGE COLUMN
def standardize_age(str_age):

        if 'years old' in str_age:
            str_age = str_age.replace('years old', '')
            output = str_age
        else:
            output=str_age
        return output
    
survey['age']=survey['age'].apply(standardize_age)
survey['age']=survey['age'].astype(int)
```

```python
##MAKE VALUES IN COLUMNS TO LOWERCASE
survey['gender']=survey['gender'].str.lower()
survey['introvert_extrovert']=survey['introvert_extrovert'].str.lower()
survey['do you play an instrument?']=survey['do you play an instrument?'].str.lower()
```

```python
##FAVORITE GENRE DATA CLEANING

def standardize_gen(str_gen):
    
        str_gen = str_gen.lower()
        str_gen = str_gen.strip()

        if '-' in str_gen:
            str_gen = str_gen.replace('-', '')
            output = str_gen
        if 'alternative rock' in str_gen or 'Alternative' in str_gen:
            str_gen = str_gen.replace('alternative rock', 'alternative')
            str_gen = str_gen.replace('Alternative', 'alternative')
            output = str_gen
        if 'oldies' in str_gen:
            str_gen = str_gen.replace('oldies', '')
            output = str_gen
        else:
            output=str_gen
        return output
    
survey['favorite_genre_1']=survey['favorite_genre_1'].apply(standardize_gen)
survey['favorite_genre_2']=survey['favorite_genre_2'].apply(standardize_gen)
survey['favorite_genre_3']=survey['favorite_genre_3'].apply(standardize_gen)
```

```python
##CLEAN HOURS BY REMOVING SYMBOLS AND ADDING AVERAGES FOR PEOPLE WHO PROVIDED RANGE
def standardize_hours(str_hours):
    str_hours=str(str_hours)

    if 'less' in str_hours or '+' in str_hours or ' hour' in str_hours or ' hrs ' in str_hours or '~' in str_hours or '?' in str_hours:
        str_hours = str_hours.replace('less than ', '')
        str_hours = str_hours.replace('+', '')
        str_hours = str_hours.replace(' hours', '')
        str_hours = str_hours.replace(' hour', '')
        str_hours = str_hours.replace(' hrs ', '')
        str_hours = str_hours.replace('~', '')
        str_hours = str_hours.replace('?', '')
        output = str_hours       
    else:
        output = str_hours
    return output

survey['listening_hours']=survey['listening_hours'].apply(standardize_hours)

def standardize_avg(str_avg):
    str_avg=str(str_avg)
    if'0-1'in str_avg:
        output='0.5'
    elif '1-2' in str_avg:
        output='1.5'
    elif '2-3' in str_avg:
        output='2.5'
    elif '3-4' in str_avg:
        output= '3.5'
    elif '5-7' in str_avg:
        output= '6'
    elif ' ' in str_avg:
        str_avg=str_avg.replace('4 ','4')
        str_avg=str_avg.replace('3 ','3')
        output=str_avg
    elif '1/3' in str_avg:
        output=np.nan
    else:
        output=str_avg
    return output
survey['listening_hours']=survey['listening_hours'].apply(standardize_avg)
survey['listening_hours']=survey['listening_hours'].astype(float)
```

```python
##ADD ' , ' TO SEPARATE THE MAJORS, SPLIT THE COLUMNS
survey['major_majors']=survey['major_majors'].replace('DSC & Business Econ','data science, business economics')
survey['major_majors']=survey['major_majors'].replace('Psychology & Human Developmental Science','psychology, human developmental science')
survey['major_majors']=survey['major_majors'].replace('Human Biology and Clinical Psychology','human biology, clinical psychology')
survey['major_majors']=survey['major_majors'].replace('CS and Math - Stats Double Major','computer science, mathematics')
survey['major_majors']=survey['major_majors'].replace('World Literature (Americas) and Sociology (Law and Society)','literature, sociology')
survey['major_majors']=survey['major_majors'].replace('Computer Science and Economics','computer science, economics')
survey['major_majors']=survey['major_majors'].replace('Psychology / Political Science','psychology, political science')
survey['major_majors']=survey['major_majors'].replace('Psychology & biochem ','psychology, biology')
   
survey[['major_1', 'major_2']] = survey['major_majors'].str.split(',', expand=True)
survey=survey.drop(['major_majors'],axis=1)

survey = survey.reindex(columns=['age', 'gender', 'state_country','major_1', 'major_2','favorite_genre_1','favorite_genre_2','favorite_genre_3','favorite_artist','listening_hours','do you play an instrument?','instrument_played','introvert_extrovert'])
```

```python
##CLEANING DATA FOR MAJORS_1
survey['major_1']=survey['major_1'].str.lower()

def standardize_maj (str_maj):
    str_maj=str(str_maj)
    str_maj=str_maj.strip()

    if 'public' in str_maj:
        output='public health'
    elif 'cog' in str_maj:
        output='cognitive science'
    elif 'psyc' in str_maj:
        output='psychology'
    elif 'econ' in str_maj:
        output='economics'
    elif 'math' in str_maj:
        output='mathematics'
    elif 'bioengineering' in str_maj:
        output= 'bioengineering'
    elif 'biol' in str_maj:
        output= 'biology'
    elif 'int' in str_maj:
        output= 'international business'
    elif 'eng' in str_maj or 'cse' in str_maj:
        output= 'engineering'
    elif 'comp' in str_maj or 'cs' in str_maj:
        output= 'computer science'
    elif 'bioin' in str_maj or 'biochem' in str_maj:
        output='biology'
    elif 'esys' in str_maj:
        output='environmental systems'
    else:
        output=str_maj
    return output

survey['major_1']=survey['major_1'].apply(standardize_maj)
```

```python
##CLEANING DATA FOR MAJOR_2
def standardize_major2 (str_maj2):
    str_maj2=str(str_maj2)
    str_maj2=str_maj2.lower()
    str_maj2=str_maj2.strip()
    if 'minor' in str_maj2 or 'none' in str_maj2 or 'machine' in str_maj2:
        output=np.nan
    elif 'econ' in str_maj2:
        output='economics'
    elif 'math' in str_maj2:
        output='mathematics'   
    elif 'bio' in str_maj2:
        output='biology'
    elif 'psyc' in str_maj2:
        output='psychology'
    else:
        output=str_maj2
    return output
survey['major_2']=survey['major_2'].apply(standardize_major2)
```

```python
##FOR LOCATION USE STATE AND COUNTRY TO INCLUDE INTERNATIONAL STUDENTS
def standardize_loc(str_loc):
    str_loc=str(str_loc)
    str_loc=str_loc.strip()
    if ' CA' in str_loc or 'California' in str_loc or 'california' in str_loc or 'CA,' in str_loc or 'San francisco' in str_loc or 'CA ' in str_loc or 'south gate' in str_loc or 'akla' in str_loc or 'Sunny' in str_loc or 'San Diego' in str_loc:
        output='california, united states'
    elif ' Texas' in str_loc or 'TX' in str_loc:
        output='texas, united states'
    elif 'WA' in str_loc or 'Wa' in str_loc:
        output='washington, united states'
    elif 'MI' in str_loc:
        output='michigan, united states'
    elif 'PA' in str_loc:
        output='pennsylvania, united states' 
    elif 'NC' in str_loc:
        output='north carolina, united states' 
    elif 'NY' in str_loc:
        output='new york, united states'
    elif 'OR' in str_loc or 'Oregon' in str_loc:
        output='oregon, united states'
    elif 'Jersey' in str_loc:
        output='new jersey, united states'
    elif 'Ohio' in str_loc:
        output='ohio, united states'
    elif 'Illinois' in str_loc:
        output='illinois, united states'
    elif 'Kansas' in str_loc:
        output='kansas, united states'  
    elif 'Arizona' in str_loc:
        output='arizona, united states' 
    elif 'Monterrey' in str_loc:
        output='nuevo leon, mexico'
    elif 'spain' in str_loc:
        output='madrid, spain' 
    elif 'Jalisco' in str_loc:
        output='jalisco, mexico' 
    elif 'Sonora' in str_loc:
        output='sonora, mexico' 
    elif 'Belo' in str_loc:
        output='minas gerais, brazil'
    else:
        output=str_loc
    return output
survey['state_country']=survey['state_country'].replace('China',np.nan)
survey['state_country']=survey['state_country'].replace('International',np.nan)
```

```python
##APPLY THE STANDARIZATION TO STATE AND COUNTRY, SEPARATE THEM INTO 2 COLUMNS
survey['state_country']=survey['state_country'].apply(standardize_loc)
survey['state_country']=survey['state_country'].replace('China',np.nan)
survey['state_country']=survey['state_country'].replace('International',np.nan)
survey['state_country']=survey['state_country'].str.lower()

survey[['state', 'country']] = survey['state_country'].str.split(',', expand=True)
survey=survey.drop(['state_country'],axis=1)

survey = survey.reindex(columns=['age', 'gender','state','country', 'major_1', 'major_2','favorite_genre_1','favorite_genre_2','favorite_genre_3','favorite_artist','listening_hours','do you play an instrument?','instrument_played','introvert_extrovert'])
```
